- shift duration: $7.5\ \mathrm{h}$
- gravitational acceleration: $9.81\ \mathrm{m/s^2}$

### Assumptions

- Each lift moves the same component through the same vertical height.
- The component starts and ends each lift at rest.
- The stated efficiency represents the overall electrical-to-useful-mechanical efficiency during lifting.
- The efficiency is assumed constant for all lifts.
- Energy used during lowering, empty return motion, standby, and auxiliary operation is neglected.
- Gravitational acceleration is constant.

### Required outputs

1. Useful mechanical energy transferred to the component during one lift, in joules
2. Electrical energy required for one lift, in joules
3. Total electrical energy for the shift, in kWh
4. Average lift-related electrical power over the complete shift, in watts

## Equation guide: from engineering equation to Python

### 1. Convert the efficiency

The lifting efficiency is the ratio of useful mechanical energy to electrical input energy:

$$
\eta = \frac{E_m}{E_e}
$$

The efficiency is given as a percentage, so it must first be converted to decimal form:

$$
\eta = \frac{\eta_{\%}}{100}
$$

where:

- $\eta_{\%}$: efficiency in percent
- $\eta$: efficiency as a decimal
- $E_m$: useful mechanical energy for one lift $[\mathrm{J}]$
- $E_e$: electrical energy required for one lift $[\mathrm{J}]$

**Python form**

```python
efficiency = efficiency_percent / 100
```

### 2. Calculate the mechanical energy for one lift

The useful mechanical energy transferred to the component is equal to its increase in gravitational potential energy:

$$
E_m = mgh
$$

where:

- $m$: component mass $[\mathrm{kg}]$
- $g$: gravitational acceleration $[\mathrm{m/s^2}]$
- $h$: lifting height $[\mathrm{m}]$
- $E_m$: mechanical energy for one lift $[\mathrm{J}]$

**Python form**

```python
mechanical_energy_per_lift = mass * g * height
```

### 3. Calculate the electrical energy for one lift

From the efficiency equation:

$$
E_e = \frac{E_m}{\eta}
$$

**Python form**

```python
electrical_energy_per_lift = (
    mechanical_energy_per_lift / efficiency
)
```

### 4. Calculate the total electrical energy for the shift

If the hoist performs $N$ identical lifts:

$$
E_s = N E_e
$$

where:

- $N$: number of lifts during the shift
- $E_s$: total electrical energy for the shift $[\mathrm{J}]$

**Python form**

```python
shift_energy_j = (
    number_of_lifts * electrical_energy_per_lift
)
```

### 5. Convert the shift energy to kilowatt-hours

The conversion between joules and kilowatt-hours is:

$$
1\ \mathrm{kWh} = 3.6 \times 10^6\ \mathrm{J}
$$

Therefore:

$$
E_{\mathrm{kWh}}
=
\frac{E_s}{3.6 \times 10^6}
$$

**Python form**

```python
shift_energy_kwh = shift_energy_j / 3_600_000
```

### 6. Convert the shift duration to seconds

Power in watts uses seconds:

$$
t_s = 3600 t_h
$$

where:

- $t_h$: shift duration $[\mathrm{h}]$
- $t_s$: shift duration $[\mathrm{s}]$

**Python form**

```python
shift_duration_s = shift_duration_h * 3600
```

### 7. Calculate the average electrical power

Average electrical power over the complete shift is:

$$
P_{\mathrm{avg}} = \frac{E_s}{t_s}
$$

where:

- $P_{\mathrm{avg}}$: average lift-related electrical power $[\mathrm{W}]$

**Python form**

```python
average_power_w = shift_energy_j / shift_duration_s
```

**Note:** This is the lift-related electrical energy averaged over the complete shift. It is not the instantaneous, peak, or rated motor power.

## Exercise 1.2 — Balancing a Machine Arm

A horizontal machine arm carries a tool on the right side of a pivot. The arm's center of mass is also located on the right side of the pivot.

A counterweight is placed on the left side of the pivot to produce an opposing moment and keep the arm in rotational equilibr



<img src="./EX2.png" width="700">


Given

- tool mass: $18\ \mathrm{kg}$
- distance from the pivot to the tool center of mass: $0.75\ \mathrm{m}$
- arm mass: $12\ \mathrm{kg}$
- distance from the pivot to the arm center of mass: $0.40\ \mathrm{m}$
- distance from the pivot to the counterweight center of mass: $0.55\ \mathrm{m}$
- gravitational acceleration: $9.81\ \mathrm{m/s^2}$

### Required outputs

1. Magnitude of the moment about the pivot caused by the tool weight
2. Magnitude of the moment about the pivot caused by the arm weight
3. Total clockwise moment about the pivot
4. Counterweight mass required for rotational equilibrium

## Physical reasoning

The tool and the arm center of mass are located on the right side of the pivot.

Their weights act vertically downward and produce **clockwise moments** about the pivot.

The counterweight is located on the left side of the pivot. Its weight acts vertically downward and produces a **counterclockwise moment**.

The pivot provides the reaction forces required to support the system. Because these reaction forces act through the pivot, they produce zero moment about the pivot.

For rotational equilibrium, the clockwise and counterclockwise moments must balance:

$$
M_c = M_t + M_a
$$

where:

- $M_c$: moment produced by the counterweight
- $M_t$: moment produced by the tool weight
- $M_a$: moment produced by the arm weight

## Assumptions

- The arm is horizontal and rigid.
- The pivot is modeled as an ideal frictionless pin.
- The pivot can provide reaction forces but does not provide a resisting moment.
- All gravitational forces act vertically downward.
- The tool is treated as a point mass located at its center of mass.
- The counterweight is treated as a point mass located at its center of mass.
- The arm's entire weight is represented as acting at its center of mass.
- The tool and arm center of mass are located on the right side of the pivot.
- The counterweight is located on the left side of the pivot.
- Other external forces and moments are neglected.

## Equation guide: from engineering equation to Python

### 1. Calculate the weight forces

The magnitude of the gravitational force acting on a mass is:

**Engineering formula**

$$
F_g = mg
$$

For the tool:

$$
F_t = m_t g
$$

For the arm:

$$
F_a = m_a g
$$

where:

- $m_t$: tool mass $[\mathrm{kg}]$
- $m_a$: arm mass $[\mathrm{kg}]$
- $g$: gravitational acceleration $[\mathrm{m/s^2}]$
- $F_t$: tool weight force $[\mathrm{N}]$
- $F_a$: arm weight force $[\mathrm{N}]$

**Python form**

```python
tool_weight_force = tool_mass * g
arm_weight_force = arm_mass * g
```

### 2. Calculate the moment produced by each weight

The general magnitude of a moment about a pivot is:

**Engineering formula**

$$
M = rF\sin(\theta)
$$

where:

- $M$: moment magnitude $[\mathrm{N\,m}]$
- $r$: distance from the pivot to the point where the force acts $[\mathrm{m}]$
- $F$: force magnitude $[\mathrm{N}]$
- $\theta$: angle between the position vector and the force

The arm is horizontal and the gravitational forces act vertically downward. Therefore:

$$
\theta = 90^\circ
$$

Since:

$$
\sin(90^\circ)=1
$$

the moment magnitude becomes:

$$
M=rF
$$

For the tool:

$$
M_t = F_t r_t
$$

For the arm:

$$
M_a = F_a r_a
$$

where:

- $r_t$: distance from the pivot to the tool center of mass $[\mathrm{m}]$
- $r_a$: distance from the pivot to the arm center of mass $[\mathrm{m}]$
- $M_t$: moment produced by the tool weight $[\mathrm{N\,m}]$
- $M_a$: moment produced by the arm weight $[\mathrm{N\,m}]$

**Python form**

```python
tool_moment = tool_weight_force * tool_distance
arm_moment = arm_weight_force * arm_center_distance
```

### 3. Calculate the total clockwise moment

The tool and the arm produce moments in the same rotational direction, so their moment magnitudes are added:

**Engineering formula**

$$
M_{\mathrm{clockwise}}
=
M_t + M_a
$$

**Python form**

```python
total_clockwise_moment = tool_moment + arm_moment
```

### 4. Calculate the counterweight mass

Take counterclockwise moments as positive.

For rotational equilibrium about the pivot:

$$
\sum M_{\mathrm{pivot}} = 0
$$

For this geometry:

$$
M_c - M_t - M_a = 0
$$

Therefore:

$$
M_c = M_t + M_a
$$

The moment produced by each weight is:

$$
M = mgr
$$

Therefore:

$$
m_c g r_c
=
m_t g r_t + m_a g r_a
$$

Because $g$ appears in every term, it cancels:

$$
m_c r_c
=
m_t r_t + m_a r_a
$$

Solving for the counterweight mass gives:

$$
m_c
=
\frac{m_t r_t + m_a r_a}{r_c}
$$

where:

- $m_c$: counterweight mass $[\mathrm{kg}]$
- $m_t$: tool mass $[\mathrm{kg}]$
- $m_a$: arm mass $[\mathrm{kg}]$
- $r_c$: distance from the pivot to the counterweight center of mass $[\mathrm{m}]$
- $r_t$: distance from the pivot to the tool center of mass $[\mathrm{m}]$
- $r_a$: distance from the pivot to the arm center of mass $[\mathrm{m}]$

**Python form**

```python
counterweight_mass = (
    tool_mass * tool_distance
    + arm_mass * arm_center_distance
) / counterweight_distance
```

### Pseudocode

```text
BEGIN

    SET tool_mass = 18
    SET tool_distance = 0.75

    SET arm_mass = 12
    SET arm_center_distance = 0.40

    SET counterweight_distance = 0.55
    SET g = 9.81

    SET tool_weight_force =
        tool_mass * g

    SET arm_weight_force =
        arm_mass * g

    SET tool_moment =
        tool_weight_force * tool_distance

    SET arm_moment =
        arm_weight_force * arm_center_distance

    SET total_clockwise_moment =
        tool_moment + arm_moment

    SET counterweight_mass =
        (tool_mass * tool_distance
        + arm_mass * arm_center_distance)
        / counterweight_distance

    DISPLAY tool_moment
    DISPLAY arm_moment
    DISPLAY total_clockwise_moment
    DISPLAY counterweight_mass

END
```

### Physical check

For rotational equilibrium, the counterweight moment must equal the total clockwise moment:

$$
M_c = M_{\mathrm{clockwise}}
$$

The calculated counterweight mass must be positive.

Because the counterweight acts at a finite distance from the pivot, changing its distance should change the required mass:

- moving the counterweight farther from the pivot should reduce the required counterweight mass;
- moving the counterweight closer to the pivot should increase the required counterweight mass.
