En textsträngar (typ `str`) kan ses som en följd (sequence) av enskilda tecken (`str` av längd 1).
`len(s)` ger längden (antalet tecken) i strängen `s`.

```python
len('Monty Python')
```

Vill man plocka ut ett enskilt tecken kan man göra det med ett *index*. Första tecknet numreras med `0` osv. Det går också att räkna bakifrån med negativa tal, sista har då index `-1` (eller ett mindre än strängens längd).

```python
'Monty Python'[1], 'Monty Python'[-3]
```

#### Mer om objekt, uttryck och variabler

En variabel är en namngiven referens till ett objekt. Ett objekt har en viss typ och kan identifieras med ett id-nummer.

Man kan byta ut objektet som en variabel refererar till.

Om ett objekt inte längre har några referenser kan det automatiskt städas bort.

```python
a = 'Monty' # Vi sätter variabeln a till att referera till objektet 'Monty' av typ str, variabelns värde
print(a, id(a), type(a))
a = 'Python' # Vi byter värde på a som nu refererar till ett annat str-objekt, 'Python'.
print(a, id(a), type(a))
```

```python
a = 'Python'
b = a
print(a, b, id(a), id(b))
a = 'Monty'
a = a + b  # Vi byter värde på a som nu refererar till värdet som skapats av uttrycket a + b.
print(a, id(a), type(a))
a += '!' # Kortform av a = a + '!', inte en förändring av objektet utan det skapas ett nytt.
print(a, id(a), type(a))
```


Objekt kan vara oföränderliga *immutable* (ex `int`, `float`, `str`, `tuple`) eller föränderliga *mutable* (ex `list`, `dict`, `set`).

För föränderliga objekt kan man ändra data i dem, till skillnad från oföränderliga.

#### Strängar är *immutable*

Strängar går inte att *ändra* på, det går till exempel inte att lägga till, ta bort eller byta ut tecken. De är, liksom talobjekt, oföränderliga (*immutable*)

Däremot kan man låta en variabel byta värde genom att låta den referera till en nytt strängobjekt, som kan vara baserat på det tidigare, på samma sätt som med tal.

```python
a = 'monty'
a[0] = 'M'  #Funkar det här?
```

#### Slice me nice

Man kan plocka ut längre delsträngar med en *slice*: `s[start: slut]` ger delsträngen som startar på index `start` och slutar *innan* index `slut`. Tänk som att indexen står mellan tecknen.

<sub>0</sub>M<sub>1</sub>o<sub>2</sub>n<sub>3</sub>t<sub>4</sub>y<sub>5</sub> ... </sub>t<sub>-3</sub>h<sub>-2</sub>o<sub>-1</sub>n

`s[start:]` ger delsträngen från och med index `start` till strängens slut. 

Vad tror du `s[:slut]` ger? Ex: `'Monty'[:3]`

Vad är resultatet av `s[:n] + s[n:]` (om `0<=n<len(s)`)?

```python
s = 'python'
s[0] = s[0].upper() # Ger fel, eftersom s är *immutable* (oföränderlig)
# s = s[0].upper() + s[1:]
print(s)
```

Låt oss bygga en funktion för att leta igenom en sträng efter ett givet tecken på tre sätt.

`search_in_string_for(s, search_for)`

Funktionen ska skriva ut ett meddelande för varje förekomst av `search_for`, skriva ut totala antalet förekomster och slutligen returnera antalet förekomster.

```python
search_in_string_for('Monty Python', 'y')
```

```python
def search_in_string_for(s, search_for):
    search_count = 0
    # Vi loopar över indexen till s
    for k in range(len(s)):
        if s[k] == search_for:
            print(f'{search_for} på plats {k}')
            search_count += 1
    print(f'Totalt {search_count} förekomster av {search_for} i {s}.')
    return search_count

# Testkörning
search_in_string_for('Monty Python', 'y')
```


Testa att själv: Bygga om `search_in_string_for` genom att använda `.find(search_for, startpos)`

(För loopen kan du (till exempel) använda `while True:` och `break` när resultatet från find är `-1`)

```python
def search_in_string_for(s, search_for):
    search_count = 0
    startpos = 0
    while True:
        k = s.find(search_for, startpos)
        if k < 0:
            break
        else:
            search_count += 1
            print(f'{search_for} på plats {k}')
            startpos = k + 1

    print(f'Totalt {search_count} förekomster av {search_for} i {s}.')
    return search_count

# Testkörning
search_in_string_for('Monty Python', 'y')
```

### Läsa filer.
Data lagras i sekundärminne som filer i ett filsystem, och kan skrivas och läsas från program. Datafiler kan vara text, eller annan data lagrad på olika sätt. Vi tittar här på textfiler, där varje tecken kodas och lagras som en eller flera *byte*. När vi läser en fil som text läser vi in den i helhet eller i delar som en textsträng.

Antag att vi har en textfil `text.txt` i samma mapp som vårt programscript. Vi kan då läsa in hela texten som en textsträng med följande kod.

```python
with open('text.txt', encoding = 'utf-8') as f:
    text = f.read()
```

Notera att konstruktionen skiljer sig lite från exemplen i Think Python.

`with open(...) as f:` med läsningen i ett block i stället för bara `f = open(...)` ser till att operativsystemet hanterar filen på ett mer säkert sätt.

Varje radbrytning är i filen ett radbrytningstecken (ascii/unicode kodnr 10).
Ett radbrytningstecken kan i python skrivas som '\n'

```python
 print(ord('\n'))
```

```python
antal_radbrytningar = 0
for c in text:
    if c == '\n':
        antal_radbrytningar += 1
print(antal_radbrytningar)
```

För stora filer vill man kanske inte läsa in hela filen på en gång. En textfil kan läsas rad för rad så här.

```python
# Läser in en fil en rad i taget och skriver ut den.
with open('text.txt', encoding = 'utf-8') as f:
    for line_of_text in f:
        print(line_of_text, end='') # end='' för att inte skriva extra ny rad
```

```python
# Läser in en fil en rad i taget, delar på orden och söker efter ett givet ord
def count_word_in_file(fileobject, word_to_count):
    count = 0
    for line_of_text in fileobject:
        print(line_of_text,end='')
        for w in line_of_text.strip().split():
            if word_to_count.lower() in w.lower(): # sant om word_to_count är en del av w
                count += 1
    return count

with open('parrot_sketch.txt', encoding = 'utf-8') as f:
    print(count_word_in_file(f, 'parrot'))
```
