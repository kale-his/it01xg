```python
import turtle
turtle.delay(0)
def koch(length):
    if length < 5:
        turtle.forward(length)
    else:
        koch(length/3)
        turtle.left(60)
        koch(length/3)
        turtle.right(120)
        koch(length/3)
        turtle.left(60)
        koch(length/3)

koch(500)
input('Stäng')
turtle.bye()
```
