## Exercise 9 : Fit, interpret, and inspect errors

predict the slope sign for power versus fuel consumption,
and for weight versus mpg. Do their slopes have the same unit?

The helper below is provided so you can focus on the analysis. It returns THREE values:
slope, intercept, and SSE (sum of squared errors).


```python
def fit_line(x, y):
    x_array = np.asarray(x, dtype=float)
    y_array = np.asarray(y, dtype=float)
    slope, intercept = np.polyfit(x_array, y_array, 1)
    predicted = slope * x_array + intercept
    sse = np.sum((y_array - predicted) ** 2)
    return slope, intercept, sse
```
