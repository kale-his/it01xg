Förra gången byggde vi en funktion för att iterativt beräkna närmevärden till kvadratroten ur ett positivt tal.

````python
def better(x, r):
    nytt_r = (x/r+r)/2
    return nytt_r

def roten_ur(x, r, acceptabelt_relativt_fel):
    while (abs((1-x/r**2)/2) > acceptabelt_relativt_fel):
        r = better(x,r)
    return r

x = 23
# Värdet på variabeln x utanför funktionen 
# har inget med parametern x att göra.

print(roten_ur(16, 1, 1e-12))
````


### Funktioner, parametrar och argument, lokala variabler

Om vi betraktar vår funktion roten_ur ovan så ser vi att tar tre argument.

I funktionens kod fungerar de som variabler, som fått värden från *argumenten* när funktionen anropades. Dessa variabler kallas funktionens *parametrar*.
Argumenten sätter alltså värden på parametrarna, som bara existerar inne i funktionen. Parametern x ges till exempel värdet 16 i exemplet.
Parametern r ges initialt värdet 1, men får nya värden under exekveringen.

Man kan också skapa nya variabler i funktionskoden. De är lokala för funktionen och existerar inte utanför. En lokal variabel kan ha samma namn som en global variabel, men de är *olika* variabler och inne i funktionen gäller den lokala.

Om ett variabelnamn inte använts för en lokal variabel kan det däremot åberopas för det aktuella värdet av en global variabel.

````python
def skriv_slott():
    print(slott)

slott = 'Camelot'
skriv_slott()
slott = 'Castle Anthrax'
skriv_slott()
skriv_slott # funktionen själv, ingen exekvering
````

En global variabel kan inte ändra värde om den inte deklareras som global i funktionskoden. Man bör vara restriktiv med att låta funktioner ändra globala variabler.

````python
def increase_x_by_one():
    global x # talar om att x inte ska vara lokal
    x = x + 1 # nu ändras den globala variabeln x

x = 23
increase_x_by_one()
print(x)
````

## Booleska värden
Ett booleskt värde (George Boole, 1815-1864) är sant (`True`) eller falskt (`False`).

Exempel på uttryck med booleska värden är likhets- och olikhets-jämförelser

* `a == b`, a är lika med b. I Python kan de vara olika men likvärdiga objekt, tex flyttal och heltal, `1.0 == 1` är sant.
* `a != b`, a är inte lika med b, motsatsen till ovanstående
* `a < b`, `a > b`, `a <= b`, `a >= b`,  $a<b,\ a>b,\ a\leq b,\ a\geq b$

För taltyper är jämförelsen numerisk, i Python kan talen ha olika typ men det är det numeriska värdet som jämförs.Till exempel flyttal och heltal: `1.0 == 1` är sant.

För teckensträngar gjörs jämförelser lexikografiskt (i bokstavsordning), utifrån unicode-nummer, ascii-versaler kommer före ascii-gemener.

Tal och textsträngar kan också tolkas booleskt. För tal tolkas 0 som falskt och övriga tal som sanna. Tomma textsträngar tolkas som falska och icke-tomma som sanna.

````python
print('a' > 'b',
      'b' > 'a',
      'B' > 'a',
      'Ö' > 'a')

print('Andersson' > 'Anderson')
````

## Villkorade satser: `while` och `if`

Vi har tidigare stött på satsen `while <villkor>:` som repeterar ett block av kod så länge ett uttryck är sant.

````python
n = 5
while n: # En heltal (`int`) tolkas som sant så länge det inte är noll
    print(n)
    n -= 1 # Minska n med 1. (a -= b) är likvärdigt med (a = a - b) 
````

Med `if <villkor>:` istället för `while <villkor>` exekveras blocket istället precis en gång om uttrycket är sant, annars hoppas det över.
`if`-blocket kan följas av ett `else`-block som exekveras om uttrycket har värdet `False`.

I följande exempel nedan använder vi operatorn `%`, *modulo*. 

`n % m` ger (minsta icke-negativa) resten vid heltalsdivison av n med m.

Om $n = q\cdot m + r$ där alla tal är heltal och $0\leq r < m$ så ges $r$ av `n % m` och $q$ av `n // m` (*heltalsdivision*).  

````python
for n in range(10):
    if n % 2 == 0: # n % m ger resten vid heltalsdivision av n med m
        print(n, 'är ett jämnt tal')
    else:
        print(n, 'är ett udda tal')
````

Om man har flera alternativa villkor kan man också lägga till ett eller flera `elif` emellan.

I satsen
```
if <A>:
    <block 1>
elif <B>:
    <block 2>
else:
    <block 3>
```
exekveras enbart block 1 om villkor A är sant, enbart block 2 om A är falskt men B är sant och block 3
om varken A eller B är sant. Är likvärdigt med den nästlade if/else-satsen
```
if <A>:
    <block 1>
else:
    if <B>:
        <block 2>
    else:
        <block 3>
```

````python
for n in range(6):
    r = n % 3 # resten av n vid division med 3
    q = n // 3
    if r == 0:
        print(n, 'är jämnt delbart med 3')
    elif r == 1:
        print(n, 'har rest 1 vid division med 3')
    else:
        print(n, 'har rest 2 vid division med 3')
    print(f'{n} = 3·{q} + {r}') # här används en f-sträng för formaterad utskrift
````

````python
n = 2
while n <= 20: # Här kör vi med while, kunde också använt for n in range(2,21):
    n_is_prime = True
    for m in range(2, n):
        if n % m == 0:
            n_is_prime = False
    if n_is_prime:
        print(n, 'är ett primtal')
    else:
        print(n, 'är ett sammansatt tal')
    n += 1
````

Låt oss definiera en funktion som testar om ett heltal är ett primtal (Ett heltal större än 1 som inte är jämnt delbart med något heltal mellan 1 och sig självt.)
