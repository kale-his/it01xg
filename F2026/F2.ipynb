## Exempel: Kvadratrot

Exempel: Vi vill implementera en algoritm för att beräkna närmevärden till kvadratroten av ett tal.
Givet ett positivt tal $x$ och ett utgångsvärde $r = 1$ vill vi justera $r$ så att $r$ kommer närmre $\sqrt{x}$, kvadratroten av $x$.

Vi ersätter $r$ med medelvärdet av $r$ och $x/r$: 
$$((x/r) + r)/2$$

Vi kan sedan upprepa detta för att få bättre och bättre värden.

### Algoritm i pseudokod

0. Indata: Talet som roten ska dras av, initialgissningen till närmevärde
1. Förbättra närmevärdet
2. Upprepa från 1. tills värdet anses bra nog
3. Utdata: Det senast beräknade närmevärdet

Vi testar först genom att göra upprepningen för hand.

```python
r = 1
x =10
```

```python
r = (x/r + r)/2
r
```

Vi kan bygga en *funktion* som givet x och r ger oss ett bättre värde.

Hur gör vi det i Python?

```python
def better(x, r):
    nytt_r = (x/r+r)/2
    return nytt_r

print(better(10,1))
```

```python
x = 10
r = 1
```

```python
r = better(x,r)
print(r)
```

Man kan visa att det relativa felet $$\frac{r-\sqrt{x}}{r}\approx \frac{1}{2}\left(1-\frac{x}{r^2}\right)$$
för små fel. Låt oss bygga en funktion som givet $x$ och godtagbart relativt fel ger ett närmevärde.
För $n$ siffrors noggrannhet vill vi ha ett relativt fel som är högst $\pm 10^{-(n+1)}$ för att vara på säkra sidan.
För att göra detta kan vi använda en konstruktion med `while`.

```python
''' struktur på en whileloop
while (<logiskt villkor>):
    <kod rad 1>
    <kod rad 2>
    ...
<nu har vi lämnat snurran>

# upprepar det indenterade blocket så länge villkoret är sant, därefter fortsätter 
# Observera : och indentering (standard fyra blanksteg)
'''
```

```python
def roten_ur(x, r, acceptabelt_relativt_fel):
    while (abs((1-x/r**2)/2) > acceptabelt_relativt_fel):
        r = better(x,r)
    return r
```

```python
print(roten_ur(10, 1, 0.000000000001))
```

Låt oss nu lägga till ett användargränssnitt som läser in ett tal och sedan skriver ut den beräknade kvadratroten.
Vi behöver då också översätta från en textsträng till taltyp.

```python
x = float(input("Ange talet du vill dra kvadratroten ur: "))
# input ger en textsträng av typ str som värde.
# float(...) tolkar strängen som ett tal
#  och ger ett talvärde av typ float
print('Kvadratroten ur ', x, 'är', roten_ur(x, 1, 1e-12), 'med ca 12 gällande siffror.')
```
