## Övning Caesarkrypto

Vi vill skapa en funktion som utför denna kryptering. Funktionen ska ta två argument, en teckenstäng (typ `str`) för meddelandet och en nyckel (typ `int`).

Användbara funktioner:
* `text.upper()` ger en ny textsträng där alla bokstäver i `text` ersatts med sin versala motsvarighet.
* `ord(tecken)` ger unicode-talet (en `int`) för tecknet `tecken`, A-Z har koder 65-90. `ord('A')` ger t ex `65`
* `chr(kod)` är omvändningen till ord och översätter från unicode-kod till tecken. `chr(65)` ger t ex `'A'`.
* Om `s` är en variabel som refererar till en sträng och c ett tecken gör `s += c` att värdet av `s` ersätts av en ny sträng där `c` lagts till på slutet.
