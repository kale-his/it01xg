#### Övning datastrukturer med dict


#### Problembeskrivning
Programmet ska gå igenom data för kvalitetskontroll av tillverkade produkter och generera en rapport. Indata består av mätdata för ett antal tillverkade produkter och toleranskrav för de olika mått som tagits på respektive produkt. Från dessa data kan man bedöma vilka av produkterna som håller alla toleranser.

Mätdata ges som en en lista av dataposter av typen `dict`. I varje datapost ges för en nyckel vilket typ av mått det är och ett mätvärde (se nedan och data.py). För varje mått finns en tabell över godtagbara toleranser för mätvärdena.

Toleranstabellen är en `dict` med poster indexerade med `id` för respektive mått. Värdet är i sin tur en `dict` som beskriver toleransen given som minsta godtagbara värde (`’min’`) och största godtagbara värde (`’max’`). 

Exempel: Det tas två mått, `’a’` och `’b’` med följande toleranstabell och mätdata.


````python
test_data = [ {'a': 23.5, 'b': 12.2},
             {'a': 23.7, 'b': 12.3},
             {'a': 23.1, 'b': 11.7},
             {'a': 23.2, 'b': 11.8},
             {'a': 22.4, 'b': 12.1}]

test_toleranser = { 'a': {'min': 22.6, 'max':23.4},
                   'b': {'min': 11.75, 'max':12.25} }
````

Vi ser här att för första fallet `test_data[0]` är inte värdet på 'a' inom toleransen, medan 'b' är inom toleransen. För `test_data[1]` är varken 'a' eller 'b' inom toleranserna, för `test_data[2]` endast 'a', för test_data[3] båda och för test_data[4] endast 'b'.
Det är alltså bara `test_data[3]` som ska anses godkänd.

#### Följande funktioner ska implementeras.

Funktionen `skapa_fellista(data, toleranstabell)`
ska, givet en lista över mätningar `data`, och en toleranstabell `toleranstabell`, gå igenom data och utifrån toleranstabell skapa en ny lista, där varje post är en lista (`list`) som innehåller de mått som inte uppfyller toleranserna. Om fellistan är tom är den mätta produkten alltså godkänd.

Funktionen 
`skapa_felstatistik(fellista, toleranstabell)`
tar en lista `fellista` (utdata från funktionen `skapa_fellista`) ska för varje mått i `toleranstabell` räkna antalet produkter  som inte håller sig inom toleransen och returnera en dict med antalet fel för varje mått. (För den här funktionen används `toleranstabell` bara för att få en lista av måttbeteckningar, värdena behövs inte.)

Med `test_data` och `test_toleranser` som ovan ska `skapa_fellista(test_data, test_toleranser)` returnera
```
[['a'],['a','b'],['b'],[],['a']]
```
och `skapa_felstatistik(skapa_fellista(test_data, test_toleranser), test_toleranser)` ska returnera
```
{'a': 3, 'b': 2}
```


##### Importera data från separat pythonfil
Vi har tidigare sett hur man kan läsa data från en komma eller semikolonseparerad datafil kan läsa in data och lagra dem som pythonobjekt. Data kan också importeras som pythonkod från en separat fil med `import`.

Om data finns på pythonform i en separat fil `data.py` (i samma mapp som programfilen) så kan variabler som tilldelas värden i denna fil hämtas till programfilen med `from data import ...`, där `...` är en lista av variabler.

Filen `data.py` innehåller bland annat tilldeling av variablerna `test_toleranser` och `test_data` som ovan så  `test_toleranser` och `test_data` kan importeras genom
```
from data import test_toleranser, test_data
```
eller, för att importera allt i filen
```
from data import *
```
