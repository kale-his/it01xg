### Satser (statements), objekt och variabler
Ett program hanterar data utifrån ett programflöde, som i programmet formuleras som en följd satser (statements).

Data paketeras i objekt av olika typ, till exempel heltal, flyttal (närmevärden av reella tal), text, och mer sammansatta typer. Ett objekt kan skapas på olika sätt

En variabel är en namngiven referens till ett objekt (variabelns värde).

Man kan tilldela ett värde till en variabel genom

>       <variabelnamn> = <uttryck>

När en variabel används i ett uttryck (expression) ersätts den av sitt värde.

Ett variabelnamn kan vara vilket ord som helst med vissa undantag
- Kan innehålla bokstäver, siffror och _, men inga andra tecken
- Får inte börja på en siffra
- Får inte vara ett *reserverat ord*
- Skiljer på versaler och gemener

```python
a = 23
b = 42
c = a + b
print(a,'+',b,'=',c) # print är en funktion som skriver ut de givna objekten på lämpligt sätt
```

## Funktioner
En funktion tar ett antal argument och ger ett värde.
Antalet argument kan vara från noll och uppåt.
En funktion kan också ha sidoeffekter.

Ett exempel på en funktion är `print` som har sidoeffekten att skriva ut sina argument, åtskilda med blanksteg.

```python
p = print(1,2,3,'Monty', 'Python')
```

```python
print(p)
```

Funktionen `print` ger alltid värdet `None`, som kan tolkas som att den inte ger något användbart värde)

En uppsättning matematiska funktioner och konstanter finns i modulen math, som man måste *importera* innan man kan använda dem.

```python
import math # Gör modulen math tillgänglig
v = math.pi/3
print(v, math.cos(v), math.sin(v))
```

```python
from math import * # Importera funktioner och konstanter från math som "standardfunktioner"
v = pi/3 # nu behövs inte modulprefixet math för pi.
print(v, cos(v), sin(v))
```
